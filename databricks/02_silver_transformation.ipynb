```python

#Create Dimensions for star schema: silver.dim_customer
from pyspark.sql import functions as F

customer = spark.table(
    "analytics_engineering.bronze.customer"
)

dim_customer = (
    customer
    .select(
        "CustomerID",
        "PersonID",
        "StoreID",
        "TerritoryID",
        "AccountNumber",
        "ModifiedDate"
    )
    .dropDuplicates(["CustomerID"])
)

print("Bronze customers:", customer.count())
print("Silver customers:", dim_customer.count())

display(dim_customer)
```

```python
dim_customer.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "analytics_engineering.silver.dim_customer"
    )

print("✓ silver.dim_customer created")
```

```python
product = spark.table(
    "analytics_engineering.bronze.product"
)

dim_product = (
    product
    .select(
        "ProductID",
        "Name",
        "ProductNumber",
        "Color",
        "StandardCost",
        "ListPrice",
        "DaysToManufacture",
        "ProductSubcategoryID",
        "ProductModelID",
        "SellStartDate",
        "SellEndDate",
        "DiscontinuedDate",
        "ModifiedDate"
    )
    .dropDuplicates(["ProductID"])
)

print("Bronze products:", product.count())
print("Silver products:", dim_product.count())

display(dim_product)
```

```python
dim_product.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "analytics_engineering.silver.dim_product"
    )

print("✓ silver.dim_product created")
```

```python
product_category = spark.table(
    "analytics_engineering.bronze.product_category"
)

dim_product_category = (
    product_category
    .select(
        "ProductCategoryID",
        "Name",
        "ModifiedDate"
    )
    .dropDuplicates(["ProductCategoryID"])
)

print("Categories:", dim_product_category.count())

dim_product_category.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "analytics_engineering.silver.dim_product_category"
    )

display(dim_product_category)
```

```python
product_subcategory = spark.table(
    "analytics_engineering.bronze.product_subcategory"
)

dim_product_subcategory = (
    product_subcategory
    .select(
        "ProductSubcategoryID",
        "ProductCategoryID",
        "Name",
        "ModifiedDate"
    )
    .dropDuplicates(["ProductSubcategoryID"])
)

print("Subcategories:", dim_product_subcategory.count())

dim_product_subcategory.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "analytics_engineering.silver.dim_product_subcategory"
    )

display(dim_product_subcategory)
```
