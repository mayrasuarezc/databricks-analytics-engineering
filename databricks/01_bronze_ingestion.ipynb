```python
from pyspark.sql import functions as F

tables = [
    "customer",
    "product",
    "product_category",
    "product_subcategory",
    "sales_order_detail",
    "sales_order_header"
]

for table in tables:
    df = spark.table(f"analytics_engineering.bronze.{table}")
    
    print(f"\n{'=' * 60}")
    print(f"TABLE: {table}")
    print(f"ROWS: {df.count()}")
    print("=" * 60)
    
    null_counts = df.select([
        F.count(
            F.when(F.col(c).isNull(), c)
        ).alias(c)
        for c in df.columns
    ])
    
    display(null_counts)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import (
    IntegerType,
    LongType,
    ShortType,
    DecimalType,
    DoubleType,
    FloatType
)

for table in [
    "customer",
    "product",
    "product_category",
    "product_subcategory",
    "sales_order_detail",
    "sales_order_header"
]:
    df = spark.table(f"analytics_engineering.bronze.{table}")

    print(f"\n{'=' * 60}")
    print(f"TABLE: {table}")
    print("=" * 60)

    numeric_cols = [
        field.name
        for field in df.schema.fields
        if isinstance(field.dataType, (
            IntegerType,
            LongType,
            ShortType,
            DecimalType,
            DoubleType,
            FloatType
        ))
    ]

    for col_name in numeric_cols:
        zero_count = df.filter(F.col(col_name) == 0).count()

        if zero_count > 0:
            print(f"{col_name}: {zero_count} rows with value 0")
```

```python
zero_price_products = (
    spark.table("analytics_engineering.bronze.product")
    .filter(
        (F.col("StandardCost") == 0) &
        (F.col("ListPrice") == 0)
    )
    .select("ProductID")
)

sold_zero_price_products = (
    spark.table("analytics_engineering.bronze.sales_order_detail")
    .join(
        zero_price_products,
        on="ProductID",
        how="inner"
    )
)

print(
    "Zero-price products:",
    zero_price_products.count()
)

print(
    "Sales order detail rows involving zero-price products:",
    sold_zero_price_products.count()
)
```

```python
#¿Todos los CustomerID de las ventas existen en Customer?
customers = spark.table("analytics_engineering.bronze.customer")
orders = spark.table("analytics_engineering.bronze.sales_order_header")

missing_customers = (
    orders
    .select("CustomerID")
    .distinct()
    .join(
        customers.select("CustomerID").distinct(),
        on="CustomerID",
        how="left_anti"
    )
)

print("Customer IDs in orders without a matching customer:")
print(missing_customers.count())

display(missing_customers)
```

```python
order_header = spark.table(
    "analytics_engineering.bronze.sales_order_header"
)

order_detail = spark.table(
    "analytics_engineering.bronze.sales_order_detail"
)

missing_orders = (
    order_detail
    .select("SalesOrderID")
    .distinct()
    .join(
        order_header.select("SalesOrderID").distinct(),
        on="SalesOrderID",
        how="left_anti"
    )
)

print("SalesOrderIDs in detail without a matching header:")
print(missing_orders.count())

display(missing_orders)
```

```python
products = spark.table(
    "analytics_engineering.bronze.product"
)

missing_products = (
    order_detail
    .select("ProductID")
    .distinct()
    .join(
        products.select("ProductID").distinct(),
        on="ProductID",
        how="left_anti"
    )
)

print("Product IDs in sales without a matching product:")
print(missing_products.count())

display(missing_products)
```

```python
from pyspark.sql import functions as F

order_header = spark.table(
    "analytics_engineering.bronze.sales_order_header"
)

order_detail = spark.table(
    "analytics_engineering.bronze.sales_order_detail"
)

fact_sales = (
    order_detail.alias("d")
    .join(
        order_header.alias("h"),
        F.col("d.SalesOrderID") == F.col("h.SalesOrderID"),
        "inner"
    )
    .select(
        F.col("d.SalesOrderID"),
        F.col("d.SalesOrderDetailID"),
        F.col("h.OrderDate"),
        F.col("h.CustomerID"),
        F.col("h.TerritoryID"),
        F.col("h.OnlineOrderFlag"),
        F.col("d.ProductID"),
        F.col("d.OrderQty"),
        F.col("d.UnitPrice"),
        F.col("d.UnitPriceDiscount"),
        F.col("d.LineTotal")
    )
)

display(fact_sales)
```

```python
print("Bronze sales_order_detail:", order_detail.count())
print("Silver fact_sales:", fact_sales.count())
```

```python
fact_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("analytics_engineering.silver.fact_sales")

print("✓ silver.fact_sales created")
```
